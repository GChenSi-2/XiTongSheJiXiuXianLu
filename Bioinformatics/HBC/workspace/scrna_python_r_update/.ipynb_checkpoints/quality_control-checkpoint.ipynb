```python
# Load 10x single-cell data
# Introduction to single-cell RNA-seq workshop
# Author: Harvard Chan Bioinformatics Core
# June 2026

# Load packages
import scanpy as sc
import numpy as np
import seaborn as sns
import anndata
import matplotlib.pyplot as plt
```

```python
# Create scanpy object from ctrl sample
ctrl = sc.read_10x_mtx("data/ctrl_raw_feature_bc_matrix/")
ctrl
```

```python
# Filter cells with less than 100 genes expressed
sc.pp.filter_cells(ctrl, min_genes = 100)
ctrl
```

```python
# 基因数/counts 这类阈值 → 用 filter_cells(更地道)
sc.pp.filter_cells(ctrl, min_genes=100)     # 就地修改 ctrl
sc.pp.filter_genes(ctrl, min_cells=3)       # 顺便:过滤几乎不表达的基因
```

```python
# Calculate QC metrics for each cell
ctrl.var["mt"] = ctrl.var_names.str.startswith("MT-")   # 人类线粒体基因以 MT- 开头(小鼠是 mt-)
sc.pp.calculate_qc_metrics(ctrl, qc_vars=["mt"], percent_top=None, log1p=False, inplace=True)
ctrl.obs.head()
```

```python
# 更推荐:数据驱动,而不是拍一个固定数
# 固定阈值的问题是"这个数字从哪来"。目前单细胞最佳实践(sc-best-practices)推荐基于分布的离群点检测:
# 用 MAD(中位数绝对偏差):把"高于中位数 3~5 个 MAD"的细胞判为离群,而不是硬定 5%/10%。这样阈值会自动适配每个数据集。
# 三个指标联合判断,别只看线粒体:pct_counts_mt、total_counts(总量太低=空滴/碎片)、n_genes_by_counts(基因数太少=低质量)。单看一个容易误判。
# 小提琴图看三个指标的分布
sc.pl.violin(ctrl, ["n_genes_by_counts", "total_counts", "pct_counts_mt"],
             jitter=0.4, multi_panel=True)
# 散点图:线粒体比例 vs 总量,常能看到一簇"低总量+高线粒体"的坏细胞
sc.pl.scatter(ctrl, x="total_counts", y="pct_counts_mt")
# 典型的坏细胞会在散点图右下/左上聚成一团(低 total_counts、高 pct_counts_mt),
# 阈值就画在"主群和这团之间的空隙"上,比背一个数字靠谱得多。
# 一句话总结:免疫/PBMC 起步 5%,一般组织 10–20%,
# 代谢旺盛组织可到 30–50%,snRNA 要压到 1–5%;
# 但最好是先画分布图、结合总量和基因数,用 MAD 离群法让数据自己告诉你阈值。

```

```python
plt.rcParams["font.sans-serif"] = ["Microsoft YaHei"]  # 或 "SimHei"
plt.rcParams["axes.unicode_minus"] = False              # 顺便修正负号显示为方框

# 每个细胞的基因数,升序排序
x = np.sort(ctrl.obs["n_genes_by_counts"].values)
# 第 i 个位置代表:基因数 ≤ x[i] 的细胞有 i+1 个
y = np.arange(1, len(x) + 1)

plt.figure(figsize=(12, 8))
plt.plot(x, y)
plt.xlabel("n_genes_by_counts (阈值)")
plt.ylabel("细胞数 (≤ 阈值)")
plt.title("每个细胞基因数以下的累计细胞数")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Create list of samples
sample_names = ["ctrl", "stim"]
# Initialize empty dictionary
dict_ad = dict()

# Loop over all the samples
for sample in sample_names:
    # Get the path for each sample
    path_data = f"data/{sample}_raw_feature_bc_matrix"
    print(path_data)

    # Load scanpy object
    ad = sc.read_10x_mtx(path_data)
    # Filter and add sample metadata
    sc.pp.filter_cells(ad, min_genes=100)
    ad.obs["sample"] = sample

    # Store scanpy object in the dictionary
    dict_ad[sample] = ad
# Dictionary that contains ctrl and stim scanpy objects
dict_ad
```

```python
# Pooling together the samples
merged_adata = anndata.concat(list(dict_ad.values()),
                              join = "outer",
                              label = "sample",
                              keys = sample_names,
                              index_unique = "_")

merged_adata
```
